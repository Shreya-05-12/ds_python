Write a Python program using list comprehension to create a list containing the squares of numbers from 1 to 10

```python
l=[x**2 for x in range(1,11)]
print(l)
```

Write a Python program using list comprehension to create a list containing all even numbers from 1 to 20.

```python
ev=[]
for i in range(1,21):
    if i%2==0:
        ev.append(i)
print(ev)
            
```

Write a Python program using list comprehension to extract numbers greater than 10

```python
lst=[90,40,1,2,3,4,9,93]
l=[x for x in lst if x>10]
print(l)
```

Write a Python program using list comprehension to extract all vowels from the string.

```python
s=input()
l=[x for x in s if x in 'aeiou']
print(l)
```

Write a Python program using list comprehension to find all numbers between 1 and 50 that are divisible by 3.

```python
l=[x for x in range(1,51) if x%3==0]
print(l)
```
