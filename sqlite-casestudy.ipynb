```python
def create_table():
    with sqlite3.connect("people.db") as conn:
        cur=conn.cursor()
        cur.execute('''
        create table if not exists persons(
        id integer primary key autoincrement,
        name text not null,
        city text not null,
        phone text not null)
        ''')
    print("table created")
```

```python
def insert_person():
    name1=input("enter name:")
    city1=input("enter city:")
    phone1=input("enter phone num:")
    with sqlite3.connect("people.db") as conn:
        cur=conn.cursor()
        cur.execute("insert into persons(name,city,phone) values(?,?,?)",(name1,city1,phone1))
    print("record inserted")


    

```

```python
def update_person():
    person_id=int(input("enter id to update"))
    new_name=input("enter name:")
    new_city=input("enter new city:")
    new_phone=input("enter new num:")
    with sqlite3.connect("people.db") as conn:
        cur=conn.cursor()
        cur.execute('''
        update persons
        set name=?,city=?,phone=?
        where id=?''',(new_name,new_city,new_phone,person_id))
        if cur.rowcount>0:
            print("updated successfully")
        else:
            print("not")
                    



```

```python
def delete_person():
    person_id=int(input("enter id to delete"))
    with sqlite3.connect("people.db") as conn:
        cur=conn.cursor()
        cur.execute("delete from persons where id=?",(person_id,))
    print("deleted")
        
```

```python
def view_all():
    with sqlite3.connect("people.db") as conn:
        cur=conn.cursor()
        cur.execute("select * from persons")
        r=cur.fetchall()
        for i in r:
            print(f"id: {i[0]} | name: {i[1]} | city: {i[2]} | phone: {i[3]} ")
        print()
```
