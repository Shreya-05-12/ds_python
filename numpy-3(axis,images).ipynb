```python
import numpy as np
arr=np.array([2,1,3,2,4,5,6])
print(arr)
print(arr.shape)
print(arr.ndim)
```

```python
arr2=arr[np.newaxis,:] #newaxis at row
print(arr2)
print(arr2.shape,arr2.ndim)
```

```python
arr2=arr[:,np.newaxis]#newaxis at column
print(arr2)
print(arr2.shape,arr2.ndim)
```

```python
a=np.array([1,2,3,4,5,6])
print(a.ndim,a.shape)
```

```python
b=np.expand_dims(a,axis=1)
print(b.shape,b.ndim)
b
```

```python
b=np.expand_dims(a,axis=0)
print(b.shape,b.ndim)
b
```
