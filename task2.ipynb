```python
def sec(lst):
    u=list(set(lst))
    print(u[-2])
        
```

```python
lst=[2,4,1,5,6,6,7]
sec(lst)
```

Create a list of student marks and count how many students scored above 60.

```python
def score(l):
    count=0
    for i in l:
        if i>60:
            count+=1
    return count
```

```python
l=[45 ,60, 70, 75, 33, 99]
print(score(l))
```

Create a list of numbers and replace every negative number with 0.

```python
def rep(li):
    for i in range(len(li)):
        if li[i]<0:
            li[i]=0
    return li
```

```python
li=[4 ,-3 ,-1, 9, 5, -8]
print(rep(li))
```

Create a list of numbers and create a new list containing only even numbers.

```python
def new_lst(s):
    n=[]
    for i in range(len(s)):
        if s[i]%2==0:
            n.append(s[i])
    return n
```

```python
s=[5,4,3,7,8,9,2,16]
print(new_lst(s))
```

Create a list of numbers and create a new list containing only numbers greater than 50.

```python
def gre(l):
    n=[]
    for i in range(len(l)):
        if l[i]>50:
            n.append(l[i])
    return n
```

```python
k=[70,80,40,33,77,99]
print(gre(k))
```
