```python
import sqlite3

conn = sqlite3.connect("emp1.db", timeout=30)
cur = conn.cursor()
```

```python
conn.execute("PRAGMA foreign_keys = ON")
print(cur.fetchall())
```

```python
cur=conn.cursor()
cur.execute('''
CREATE TABLE if not exists department(
    id INTEGER PRIMARY KEY,
    d_name TEXT,
    emp_id integer,
    foreign key (emp_id) references company(id)
);''')
conn.commit()

```

```python
import sqlite3

conn = sqlite3.connect("mru.db", timeout=30)
cur = conn.cursor()

conn.execute("PRAGMA foreign_keys = ON")

print("Connected successfully")
```

```python
cur.execute('''
CREATE TABLE IF NOT EXISTS department (
    id INTEGER PRIMARY KEY,
    d_name TEXT,
    emp_id INTEGER,
    FOREIGN KEY (emp_id) REFERENCES company(id)
);
''')

conn.commit()

print("Department table created")
```

```python
cur=conn.cursor()
cur.execute('''
select company.name,department.id
from company cross join department;
''')

r=cur.fetchall()
for i in r:
    print(i)
conn.close()
```

```python
cur=conn.cursor()
cur.execute('''
select company.name,company.salary,department.id
from company inner join department
on company.id=department.id;
''')

r=cur.fetchall()
for i in r:
    print(i)
conn.close()
```
